# AdamW Optimiser

```python
import numpy as np
from tqdm import tqdm

def gradient(w):
    return np.array([1000 * w[0], 0.0001 * w[1]])

lr = 0.0015
beta1, beta2, eps = 0.9, 0.999, 1e-5

w_adam = [1.0, 1.0]
w_sgd = [1.0, 1.0]

m = np.zeros(2)
v = np.zeros(2)

for step in tqdm(range(1, 600)):

    g = gradient(w_adam)

    m = beta1 * m + (1 - beta1) * g
    v = beta2 * v + (1 - beta2) * g**2

    m_hat = m / (1 - beta1**step)
    v_hat = v / (1 - beta2**step)

    w_adam = w_adam - lr * m_hat / (np.sqrt(v_hat)+eps)
    w_sgd = w_sgd - lr * gradient(w_sgd)

    if step in [1, 5, 10, 50, 100, 150, 200, 300,400, 500]:
        print("-========")
        print(f"AdamW {np.around(w_adam, 5)}")

        print(f"SGD {np.round(w_sgd, 5)}")
        print("=========")
```

# MLP From Scratch

```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

weights = np.array([1.0, 0.0])
inputs = np.array([0.5, -0.3])
bias = 0.1
lr = 0.5
target = 1.0

prediction = sigmoid(inputs @ weights + bias)
loss = (prediction - target) ** 2
print("before : ", round(prediction, 4), "loss : ", round(loss, 4))


d_weighted_sum = 2 * (prediction - target) * (prediction) * (1 - prediction)
weights -= lr * d_weighted_sum * inputs
bias -= lr * d_weighted_sum

prediction = sigmoid(inputs @ weights + bias)
loss = (prediction - target) ** 2
print("after : ", round(prediction, 4), " loss : ", round(loss, 4))
print("Weights now : ", weights.round(3))
```

```python
"""Simple XOR problem solving using MLP"""

from tqdm import tqdm

np.random.seed(0)
inputs = np.array([[0,1], [1,0], [0,0], [1,1]])
targets = np.array([[1], [1], [0], [0]])

def sigmoid(z):
    return 1 / (1 + np.exp(-z))


w_hidden = np.random.randn(2, 4)
b_hidden = np.zeros((1, 4))
w_out = np.random.randn(4, 1)
b_out = np.zeros((1, 1))

print("===========================")
print(w_hidden)
print(b_hidden)
print("===========================")
print(w_out)
print(b_out)
print("===========================")
learning_rate = 1.0

for i in range(1000):
    hidden = sigmoid(inputs @ w_hidden + b_hidden)
    output = sigmoid(hidden @ w_out + b_out)
    loss = np.mean((output - targets) ** 2)

    d_output = 2 * (output - targets) / len(targets) * output * (1 - output)
    d_hidden = (d_output @ w_out.T) * hidden * (1 - hidden)

    w_out -= learning_rate * hidden.T @ d_output
    b_out -= learning_rate * d_output.sum(axis=0, keepdims=True)

    w_hidden -= learning_rate * inputs.T @ d_hidden
    b_hidden -= learning_rate * d_hidden.sum(axis=0, keepdims=True)

    if i % 1000 == 0:
        print("step", step, "loss", round(loss, 4))

print("final predictions:", output.round(3))
```
